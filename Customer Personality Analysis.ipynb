Loading the dataset

```python
import pandas as pd

df = pd.read_csv("marketing_campaign.csv", sep="\t")
df.head()
```

```python
print(df.info())
```

Checking for missing values

```python
df.isnull().sum()
```

Handling missing values (Income)

```python
df['Income'] = df['Income'].astype(str).str.replace(',', '')
df['Income'] = pd.to_numeric(df['Income'], errors='coerce')
```

```python
df['Income'] = df['Income'].fillna(df['Income'].median())
```

Checking if the null values are filled

```python
df['Income'].isnull().sum()
```

Removing duplicates

```python
print("Duplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)
```

Cleaning column names

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

Fixing date format

```python
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')
df.head()
```

Standardizing text columns

```python
df['education'] = df['education'].str.lower().str.strip()
df['marital_status'] = df['marital_status'].str.lower().str.strip()
```

Fixing data types

```python
df['year_birth'] = df['year_birth'].astype(int)
df['income'] = df['income'].astype(float)
```

Making an age column

```python
from datetime import datetime
current_year = datetime.now().year

df['age'] = current_year - df['year_birth']
```

Removing extreme income values - Basically handling outliers in income

```python
Q1 = df['income'].quantile(0.25)
Q3 = df['income'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['income'] >= lower) & (df['income'] <= upper)]
```
